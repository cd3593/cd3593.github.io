# Advanced Computing Streamlit App

For our Advanced Computing final project, our team, Purple Flamingo, built an interactive Streamlit dashboard for analyzing U.S. electricity demand, fuel mix, grid stress, and wholesale power prices.

## Project Background

The project began with a much broader proposal. We originally wanted to study electricity demand, fuel mix changes, interregional grid behavior, and weather impacts across multiple U.S. regions by combining EIA electricity data with NOAA weather data. After receiving feedback, we realized that this scope was too ambitious for the timeline and technical complexity of the assignment.

We therefore narrowed the project into a clearer question: how does the electricity system behave during unusually high or low demand periods? This change made the project more focused, while still keeping the original policy motivation of understanding how the power system responds under stress.

## Dashboard Pages

Our final application allows users to explore this question through several connected pages:

1. The fuel type demand page shows how electricity demand changes across major generation sources over time.

2. The regional demand page compares demand patterns across balancing authorities and regions.

3. The grid stress page identifies unusually high- and low-demand days using a z-score threshold, then compares fuel mix behavior on those anomaly days against normal days.

4. The price-demand page adds wholesale electricity prices from GridStatus and compares them with EIA regional demand data for CAISO, NYISO, and ERCOT.

## Main Finding

The main finding from the project is that grid stress is easier to understand when demand, fuel mix, and price are examined together rather than separately. The app does not just show when demand is high or low; it also helps reveal which fuel sources shift during those periods and whether wholesale prices appear to move alongside demand.

## Technical Structure

From a technical perspective, the project was built as a small end-to-end data product rather than a static notebook. The logic of the code follows a clear pipeline:

1. The EIA API and GridStatus data are pulled through separate data loading scripts.

2. The data is cleaned, standardized, and stored in BigQuery.

3. The Streamlit pages query BigQuery and turn the processed data into interactive charts.

The codebase separates different responsibilities across files. The main navigation file connects the app pages, the BigQuery utility file manages database connection and SQL queries, the data utility file handles cleaning, unit conversion, daily aggregation, z-score anomaly detection, and fuel mix comparison, and the individual Streamlit pages focus on visualization and user interaction.

This structure made the project easier to maintain and helped the dashboard run more efficiently because the app reads from BigQuery instead of repeatedly calling raw APIs every time a user opens it.

## Policy Relevance

This app could be useful for policy makers because it translates complex electricity data into a more accessible decision-support tool. A policy maker could use it to identify unusual demand periods, compare how different regions behave under stress, observe whether the grid relies more heavily on certain fuel types during high-demand days, and examine how wholesale electricity prices respond in selected ISOs.

These insights could support conversations around grid reliability, energy transition planning, fuel dependence, regional vulnerability, and future investment in flexible or cleaner generation sources.
